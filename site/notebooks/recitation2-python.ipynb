## 1. Bounds describe functions

Let f(n) and g(n) be nonnegative for sufficiently large integer n, with g(n) positive there. We compare their growth as n increases, ignoring a finite initial range and constant factors.

- **f ∈ O(g):** there exist c > 0 and n₀ such that 0 ≤ f(n) ≤ c·g(n) for every n ≥ n₀. This is an upper bound.
- **f ∈ Ω(g):** there exist c > 0 and n₀ such that 0 ≤ c·g(n) ≤ f(n) for every n ≥ n₀. This is a lower bound.
- **f ∈ Θ(g):** there exist c₁, c₂ > 0 and n₀ such that 0 ≤ c₁·g(n) ≤ f(n) ≤ c₂·g(n) for every n ≥ n₀. This is a tight bound.

We often write f(n) = O(g(n)) as shorthand for membership in a set of functions; it is not ordinary equality. A tight bound matches the function's order of growth, not its exact value.

### One example, several valid statements

For f(n) = 5n + 8 and n ≥ 1, we have 5n ≤ f(n) ≤ 13n. Thus f ∈ Θ(n), using c₁ = 5, c₂ = 13, and n₀ = 1. It is also true that f ∈ O(n²), since 13n ≤ 13n² for n ≥ 1. That upper bound is valid but loose.

For a fixed logarithm base greater than 1, the following functions have strictly increasing asymptotic growth: `1, log n, n, n log n, n², n³, 2ⁿ`. Constant factors such as `100n` versus `n` do not change the Θ class. Asymptotic analysis alone does not predict which implementation is faster for a particular small input.

### Case analysis comes first

A running time depends on the input, not just its length. First choose the function: best case, worst case, or an expectation under a stated input distribution or algorithmic randomness. Then bound that function with O, Ω, or Θ.

For standard insertion sort on length n ≥ 2, the best-case time is Θ(n) on an already sorted array and the worst-case time is Θ(n²) on a reverse-sorted array with distinct keys. Both functions have upper and lower bounds. **O does not mean worst case; Ω does not mean best case.**

## 2. Count calls and combining work

1. **Divide:** form smaller subproblems and account for that work.
2. **Conquer:** recursively solve the required subproblems.
3. **Combine:** construct the full answer from their answers.
4. **Base case:** handle small inputs directly; ensure recursion makes progress.

For two equally sized recursive subproblems, a typical recurrence is `T(n) = 2T(n/2) + work outside the recursive calls`. The outside work includes dividing, combining, and other local operations. The size of a subproblem alone does not determine total time.

### Binary search: one surviving half

With constant-time array indexing and index bounds (no copied slices), worst-case binary search follows one half-sized subproblem after constant work: `B(n) = B(⌊n/2⌋) + Θ(1)` until a constant-size base case. There are Θ(log n) calls, so worst-case time is Θ(log n). A successful search that finds the target at the first midpoint takes Θ(1) time. Binary search is more precisely called *decrease-and-conquer*: only one smaller problem remains.

### Merge sort: solve both halves, then merge

For the standard implementation that merges every pair of sorted halves, assume n = 2ᵏ, n ≥ 1, and a constant-time singleton base case. Then `M(n) = 2M(n/2) + Θ(n)`. The two recursive calls sort both halves. Copying/dividing and merging together take linear work per internal node.

- At depth d, there are 2ᵈ subproblems of size n/2ᵈ.
- At each internal level, the total local work is Θ(2ᵈ · n/2ᵈ) = Θ(n).
- There are log₂ n internal levels and n constant-cost leaves.
- Total time is Θ(n log n) for n ≥ 2. Non-power-of-two sizes need floors/ceilings but have the same asymptotic bound.

No Master Theorem is needed: count the work per level and how many levels there are. Standard array merge sort also uses O(n) auxiliary storage when its merge buffers are reused across calls; the recursion stack adds O(log n).

### Trace a combine step

Merge `[1, 4, 9]` and `[2, 2, 3, 10]`. Compare only the two unconsumed heads. The output grows as `[1]`, `[1, 2]`, `[1, 2, 2]`, `[1, 2, 2, 3]`, `[1, 2, 2, 3, 4]`, `[1, 2, 2, 3, 4, 9]`. The left side is now exhausted; append the right tail `[10]`.

Choose the left head first when values tie. For integer values the two copies look identical; for tagged records this rule preserves their relative order across the left and right halves.
